## 1. ¿Qué es una función?

Una **función** es un bloque reutilizable de código que realiza una tarea específica.
Sirve para **organizar** programas, **evitar repetir código** y **dar nombres** a operaciones.


**Ventajas de usar funciones:**

- Reutilización de código
- Claridad y mantenimiento
- Pruebas más fáciles
- División de problemas grandes en partes pequeñas


---
## 2. Definir y llamar funciones (`def`)

La forma más común de crear funciones es con la palabra clave `def`:


```python
def saludar():
    print("¡Hola, mundo!")

# Llamar a la función
saludar()
```

También podemos crear funciones que acepten **parámetros** para personalizar su comportamiento.


```python
def saludar_a(nombre):
    print(f"Hola, {nombre}")

saludar_a("Kevin")
```

---
## 3. Parámetros y argumentos

- **Parámetros**: nombres en la definición de la función.
- **Argumentos**: valores concretos que pasamos al llamar la función.


```python
def area_rectangulo(base, altura):
    return base * altura

print(area_rectangulo(5, 3))     # argumentos posicionales
print(area_rectangulo(base=5, altura=3))  # argumentos nombrados
```

---
## 4. Valores de retorno (`return`)

`return` finaliza la función y devuelve un valor. Si no se especifica, la función devuelve `None`.


```python
def es_par(n):
    if n % 2 == 0:
        return True
    return False

resultado = es_par(8)
print("¿Es par?:", resultado)
```

---
## 5. Variables locales, globales y alcance (scope)

- **Local**: definida dentro de una función, solo existe allí.
- **Global**: definida fuera de cualquier función, accesible desde dentro (lectura).
- Evita abusar de variables globales.


```python
x = 10  # global

def demo_scope():
    x = 5  # local (somos libres de usar el mismo nombre localmente)
    print("Dentro de la función, x =", x)

demo_scope()
print("Fuera de la función, x =", x)
```

## 6. Documentar funciones (docstrings) y `help()`

Usamos **docstrings** (cadenas triple-comilla en la primera línea del cuerpo) para describir qué hace la función, sus parámetros y su retorno.


```python
def promedio(a, b):
    """
    Calcula el promedio de dos números.

    Parámetros:
        a (float | int): Primer número
        b (float | int): Segundo número

    Retorna:
        float: Promedio de a y b
    """
    return (a + b) / 2

help(promedio)
print("Promedio:", promedio(10, 20))
```

## 7. Parámetros por defecto y argumentos nombrados

Podemos asignar valores **por defecto** a los parámetros. Esto permite omitirlos al llamar.


```python
def potencia(base, exponente=2):
    return base ** exponente

print(potencia(4))        # usa exponente=2 por defecto
print(potencia(4, 3))     # pasa otro valor
print(potencia(exponente=4, base=2))  # argumentos nombrados
```

## 8. `*args` y `**kwargs` (parámetros variables)

- `*args` captura argumentos posicionales variables en una **tupla**.
- `**kwargs` captura argumentos nombrados variables en un **diccionario**.


```python
def resumen(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

resumen(1, 2, 3, nombre="Ana", activo=True)
```

## 9. Funciones anónimas (`lambda`) y funciones como datos

`lambda` crea funciones pequeñas de una sola expresión. En Python, las funciones son **ciudadanos de primera clase**: se pueden guardar en variables, pasar como argumento, etc.


```python
doble = lambda x: x * 2  # función anónima
print(doble(7))

def aplicar(funcion, valor):
    return funcion(valor)

print(aplicar(doble, 10))
```

## 10. Funciones integradas útiles relacionadas

- `len`, `sum`, `min`, `max`
- `map`, `filter`
- `sorted` con `key=`


```python
nums = [3, 1, 4, 1, 5, 9]
print("len:", len(nums))
print("sum:", sum(nums))

# map y filter
cuadrados = list(map(lambda x: x*x, nums))
pares = list(filter(lambda x: x % 2 == 0, nums))
print("cuadrados:", cuadrados)
print("pares:", pares)

# sorted con key
palabras = ["python", "funcion", "lambda", "args"]
print(sorted(palabras, key=len))
```

## 11. Anotaciones de tipo (type hints) básicas

Las **anotaciones** ayudan a la lectura y a herramientas externas (no cambian la ejecución).


```python
from typing import Iterable, List

def total(valores: Iterable[int]) -> int:
    suma = 0
    for v in valores:
        suma += v
    return suma

print(total([1, 2, 3]))
```

## 12. Pruebas simples con `assert` y `doctest`

- `assert` permite verificar condiciones rápidamente.
- `doctest` ejecuta ejemplos escritos en el docstring.


```python
def factorial(n):
    """
    Calcula el factorial de n (n!).

    Ejemplos (doctest):
    >>> factorial(0)
    1
    >>> factorial(5)
    120
    """
    assert n >= 0, "n debe ser >= 0"
    resultado = 1
    for i in range(2, n+1):
        resultado *= i
    return resultado

if __name__ == "__main__":
    import doctest
    doctest.testmod(verbose=True)
    print("factorial(6) =", factorial(6))
```

## 13. Recursión (idea básica)

Una función **recursiva** se llama a sí misma. Debe tener caso base para terminar.


```python
def cuenta_regresiva(n):
    if n <= 0:
        print("¡Despegue!")
    else:
        print(n)
        cuenta_regresiva(n - 1)

cuenta_regresiva(5)
```

## 15. Soluciones propuestas

```python
def celsius_a_fahrenheit(c):
    return c * 9/5 + 32

print(celsius_a_fahrenheit(0), celsius_a_fahrenheit(100))
```

```python
def max_de_tres(a, b, c):
    mayor = a
    if b > mayor:
        mayor = b
    if c > mayor:
        mayor = c
    return mayor

print(max_de_tres(3, 9, 1))
```

```python
def es_palindromo(texto):
    limpio = "".join(ch.lower() for ch in texto if not ch.isspace())
    return limpio == limpio[::-1]

print(es_palindromo("Anita lava la tina"))
print(es_palindromo("Hola mundo"))
```

```python
def estadisticas(nums):
    assert len(nums) > 0, "La lista no puede estar vacía"
    ordenados = sorted(nums)
    n = len(nums)
    media = sum(nums) / n
    # mediana
    if n % 2 == 1:
        mediana = ordenados[n // 2]
    else:
        mediana = (ordenados[n//2 - 1] + ordenados[n//2]) / 2
    return {"media": media, "mediana": mediana}

print(estadisticas([1, 2, 3, 4, 5]))
print(estadisticas([10, 2, 7, 4]))
```

```python
def tabla(datos, alineacion="izq"):
    # calcular anchos
    if not datos:
        return ""
    num_cols = max(len(fila) for fila in datos)
    anchos = [0] * num_cols
    for fila in datos:
        for i, celda in enumerate(fila):
            anchos[i] = max(anchos[i], len(str(celda)))
    # función de alineación
    def alinear(texto, ancho):
        texto = str(texto)
        if alineacion == "der":
            return texto.rjust(ancho)
        elif alineacion == "centro":
            izq = (ancho - len(texto)) // 2
            der = ancho - len(texto) - izq
            return " " * izq + texto + " " * der
        else:
            return texto.ljust(ancho)
    # construir líneas
    lineas = []
    for fila in datos:
        celdas = [alinear(c, anchos[i]) for i, c in enumerate(fila)]
        lineas.append(" | ".join(celdas))
    return "
".join(lineas)

muestra = [
    ["Nombre", "Edad", "Ciudad"],
    ["Ana", "23", "San José"],
    ["Kevin", "28", "Cartago"],
]
print(tabla(muestra, "izq"))
print()
print(tabla(muestra, "der"))
print()
print(tabla(muestra, "centro"))
```
